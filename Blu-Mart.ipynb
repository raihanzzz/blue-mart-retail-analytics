Data Analysis & Visualisation

Q1. How the sales are changing over time ?

```python
monthly_sales = df.groupby(df['date'].dt.to_period('M'))['total_value'].sum()
monthly_sales
```

```python
monthly_sales.index = monthly_sales.index.astype(str)
monthly_sales
```

```python
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o',color='purple')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Q2. Which categories generates the highest sales and profit ?

```python
#profit = total value - quantity * cost price

df['profit'] = df['total_value'] - (df['quantity'] * df['cost_price'])
category_perf = df.groupby('category').agg(total_sales=('total_value', 'sum'),
                           total_profit=('profit', 'sum')).sort_values('total_sales',ascending=False)
category_perf

#visulatization--bar chart

x=np.arange(len(category_perf))
width=0.35

plt.figure(figsize=(11,5))
plt.bar(x-width/2,category_perf['total_sales'],width=width,label='Total Sales', color='blue')
plt.bar(x+width/2,category_perf['total_profit'],width=width,label='Total Sales', color='green')
plt.xticks(x,category_perf.index, rotation=90)
plt.ylabel('Amount')
plt.xlabel('Catagory')
plt.title('Catagory wise Sales vs profit')
plt.legend()
plt.tight_layout()
plt.show()
```

Q3. Which products are the most profitable

```python
product_perf=df.groupby('subcategory').agg(total_sales=('total_value', 'sum'),
                                           total_profit=('profit','sum'))
product_perf=product_perf.sort_values('total_profit', ascending=False)
top10_profitable = product_perf.head(10)
top10_profitable
```

```python
#Visualizatio--bar chart
plt.figure(figsize=(11,5))
plt.bar(top10_profitable.index,top10_profitable['total_profit'],color='red')
plt.xlabel('Product')
plt.ylabel('TotalProfit')
plt.title('Top 10 most profitable products')
plt.xticks(rotation=90)
plt.show()
```

Q1. AD-Hoc Which high sales product are at risk of stock out ?

```python
product_sales=df.groupby('sku_id')['quantity'].sum().reset_index()
product_sales
```

```python
risk_df=product_sales.merge(idf,on='sku_id',how='left')
risk_df
```

```python
high_sales_low_stocks = risk_df[(risk_df['quantity']>risk_df['quantity'].median()) &
    (risk_df['stock_on_hand']< risk_df['stock_on_hand'].median())] 
```

```python
# High sales low stock --> potential stockout risk
```

```python
high_sales_low_stocks
```

Q3. AD-Hoc Which product have high sales but low profit ?

```python
product_summary=df.groupby('sku_name').agg(total_sales=('total_value','sum'),
                                           total_profit=('profit','sum'))
product_summary
```

```python
high_sales_low_profit = product_summary.loc[
    (product_summary['total_sales'] > product_summary['total_sales'].median()) &
    (product_summary['total_profit'] < product_summary['total_profit'].median())
]
high_sales_low_profit
```
